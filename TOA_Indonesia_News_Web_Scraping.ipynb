```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd

# Load the website URL
url = "https://toa.co.id/news"

# 1. Initialize the Chrome driver
driver = gs.Chrome()

try:
    # 2. Open the webpage
    driver.get(url)

    # 3. Wait up to 10 seconds for the news list container to load
    wait = WebDriverWait(driver, 10)
    wait.until(EC.presence_of_element_located((By.CLASS_NAME, "rn-dl__Horizon")))

    # 4. Find all news containers using a CSS Selector (ideal for handling multiple classes)
    daftar_berita = driver.find_elements(By.CSS_SELECTOR, 'dl.rn-dl__Horizon.is-column3.rn-linkBar--smp')

    news_data = []

    # 5. Loop through each element container and extract the text
    for berita in daftar_berita:
        # Find sub-elements within the parent container
        tgl_berita = berita.find_element(By.CSS_SELECTOR, "span.rn-day").text
        nama_berita = berita.find_element(By.TAG_NAME, "a").text.strip()

        news_data.append({"Date": tgl_berita, "News Title": nama_berita})

    # 6. Process data into a DataFrame
    if news_data:
        df_news = pd.DataFrame(news_data)

        # Kept your original Jupyter Notebook display sorting
        display(df_news.sort_values(by="Date", ascending=True).head())
    else:
        print("No data collected.")

except Exception as e:
    print(f"Website is error or elements couldn't be loaded: {e}")

finally:
    # 7. Close the browser session safely
    driver.quit()
```
